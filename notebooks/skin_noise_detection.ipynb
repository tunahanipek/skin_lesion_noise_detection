# Skin and Noise Detection Project

This notebook is dedicated to the exploratory data analysis, model training, and evaluation for the skin and noise detection project. The goal of this project is to develop a robust model that can accurately detect skin regions and distinguish noise in images. 

## Project Structure
- `data/train`: Contains the training dataset for skin and noise detection.
- `data/test`: Contains the testing dataset for evaluating the model's performance.
- `data/validation`: Contains the validation dataset for tuning the model's hyperparameters.
- `src/preprocessing.py`: Functions for data preprocessing, including loading datasets, data augmentation, and normalization.
- `src/model.py`: Defines the model architecture for skin and noise detection, including training and evaluation logic.
- `src/utils.py`: Utility functions for logging, metrics calculation, and visualization.
- `requirements.txt`: Lists the Python packages required for the project.
- `README.md`: Overview of the project, setup instructions, and usage guidelines.
